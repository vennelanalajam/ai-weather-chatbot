```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


data = {"c":20,"python":100,"java":5,"NoSql":30}
keys_list = list(data.keys())
values_list = list(data.values())

fig = plt.figure(figsize = (10,5))
plt.plot(keys_list,values_list,color="r",marker = "*",linestyle="--")
plt.xlabel("SUBJECTS")
plt.ylabel("STUDENTS")
plt.title("Student enrolled in different courses")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


data = {"c":20,"python":100,"java":5,"NoSql":30}
keys_list = list(data.keys())
values_list = list(data.values())

fig = plt.figure(figsize = (5,5))
plt.bar(keys_list,values_list,color="g")
plt.xlabel("SUBJECTS")
plt.ylabel("STUDENTS\n")
plt.title("Student enrolled in different courses\n\n\n")
plt.show()
```

# Boxplot():

detecting the outliers
for example x = {1,2,3,4,5,6,7,100}
in this the value 100 will be different for others soo that will be removed in this box plot

```python
x = np.random.normal(20,1,165)
y = np.random.normal(20,2,165)
z = np.random.normal(20,3,165)
data = [x,y,z]
plt.boxplot(data)
plt.show()
```

# pie plot() with diffirent



```python
labels_dmart = ["fruits","veggitables","grosery","snacks","cloths"]
sizes = [300,250,500,600,500]
colors = ["green","pink","blue","yellow","red"]
explode = [0,0,0,0,0.1]


plt.pie(sizes,colors = colors,labels=labels_dmart,autopct='%.2f%%',explode=explode)
plt.title("DMart review")
plt.show()
```
